# Notebook Forense - Análisis de **un episodio**
- Unidad narrativa: **intervalo de decisión** → cadena causal (dinámica → control → métricas/reward → decisión RL → update).
- Fuente: **JSON en chunks de episodios**.

## Caja 0 - Imports + settings (usuario)

```python
from pathlib import Path
import json, re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------
# USER INPUTS
# -------------------------
RUN_DIR = Path(r".../results_history_CartPole/cart_pole/20260219_1514")  # <-- input
EPISODE_ID = 0                                                           # <-- input
INTERVAL_ID = 0                                                          # <-- input

# ventanas / reducción
WINDOW_N = 60
N_UNIFORM = 25
TOPK_L = 3

# preferencia raw/norm
USE_RAW_STATE = True
```

## Caja 1 - Helpers mínimos

```python
def extract_episode_id(ep):
    if isinstance(ep, dict) and 'episode_id' in ep:
        try:
            return int(ep['episode_id'])
        except Exception:
            return None
    if isinstance(ep, dict) and isinstance(ep.get('episode_data'), dict) and 'episode_id' in ep['episode_data']:
        try:
            return int(ep['episode_data']['episode_id'])
        except Exception:
            return None
    return None


def load_json(path: Path):
    with open(path, 'r', encoding='utf-8') as f:
        return json.load(f)


def discover_chunk_files(run_dir: Path):
    # Heurística simple: toma *.json y excluye metadata/config por nombre
    files = []
    for p in sorted(run_dir.resolve().rglob('*.json')):
        name = p.name.lower()
        if 'metadata' in name:
            continue
        if 'config' in name:
            continue
        files.append(p)
    return files


#def build_episode_index(chunk_paths):                  ### SE ELIMINA YA QUE ES MUY COSTOSO ESTE MÉTODO, RESUELTO MANUAL
#    # episode_id -> (chunk_path, index_in_chunk)
#    idx = {}
#    for cp in chunk_paths:
#        try:
#            obj = load_json(cp)
#        except Exception:
#            continue
#        episodes = list(iter_episode_dicts(obj))
#        for i, ep in enumerate(episodes):
#            eid = extract_episode_id(ep)
#            if eid is None:
#                continue
#            if eid not in idx:
#                idx[eid] = (cp, i)
#    return idx

def find_key_by_value(episode_index: dict, episode_id:int):
    for file_name, ep_groups in episode_index.items():
        if episode_id in ep_groups:
            return file_name
        return None
        
def load_episode_from_index(episode_index: dict, episode_id: int):
    dir_file = find_key_by_value(episode_index, episode_id)
    episodes = list(load_json(dir_file))
    ep_data = episodes[episode_id]
    return ep_data, dir_file


def list_series_keys(d):
    return [k for k, v in d.items() if isinstance(k, str) and isinstance(v, list)]


def build_df(series_dict):
    keys = list_series_keys(series_dict)
    if not keys:
        return pd.DataFrame()

    lists = {k: series_dict[k] for k in keys if isinstance(series_dict.get(k), list)}
    if not lists:
        return pd.DataFrame()

    n = max(map(len, lists.values()))
    data = {k: v + [np.nan] * (n - len(v)) for k, v in lists.items()}
    return pd.DataFrame(data)


def pick_first_existing(columns, candidates):
    for k in candidates:
        if k in columns:
            return k
    return None


def absmax(s):
    if s is None or len(s) == 0:
        return None
    a = np.abs(pd.to_numeric(pd.Series(s), errors='coerce')).dropna()
    return float(a.max()) if len(a) else None


def window_indices(center, window_n, lo, hi):
    half = window_n // 2
    a = max(lo, center - half)
    b = min(hi, center + half)
    return list(range(a, b + 1))


def topk_L(interval_row: dict, k=3, prefix='L_'):
    items = []
    for name, val in interval_row.items():
        if not isinstance(name, str) or not name.startswith(prefix):
            continue
        try:
            v = float(val)
        except Exception:
            continue
        items.append((name, v, abs(v)))
    items.sort(key=lambda x: x[2], reverse=True)
    return [{'name': n, 'value': v} for n, v, _ in items[:k]]
```

## Caja 2 - Discover chunks + index de episodios

```python
chunk_files = discover_chunk_files(RUN_DIR)
print(f'chunks encontrados: {len(chunk_files)}')
chunk_files[:5]
```

```python
n_episodes = 5000
ep_for_file = int(n_episodes/len(chunk_files))
episode_index = {}
idx_i = 0
idx_f = 0
id_dir_file = 0
for cf in chunk_files:
    idx_f += ep_for_file
    episode_index[cf] = list(range(idx_i, idx_f))
    idx_i = idx_f
    id_dir_file += 1
list(episode_index.items())[:2]
```

## Caja 3 - Cargar episodio

```python
episode, source_chunk = load_episode_from_index(episode_index, EPISODE_ID)
print('source_chunk:', source_chunk)
print('episode keys:', list(episode.keys()))

step_data = episode.get('step_data', {})
interval_data = episode.get('interval_data', {})
end_data = episode.get('end_episode_data', {})

print('step_data keys:', len(step_data))
print('interval_data keys:', len(interval_data))
print('end_data keys:', len(end_data))
```

## Caja 4 - DataFrames step/interval (alinear longitudes)

```python
step_df = build_df(step_data)
interval_df = build_df(interval_data)

print('step_df shape:', step_df.shape)
step_df.head(3)
```

## Caja 5 — Ficha del episodio (hechos mínimos)

```python
terminated = bool(end_data.get('end_terminated', False))
termination_reason = end_data.get('end_termination_reason', '')

n_steps = len(step_df) if len(step_df) else int(step_data.get('n_step', 0) or 0)
n_intervals = len(interval_df) if len(interval_df) else int(interval_data.get('n_interval', 0) or 0)

sat_key = pick_first_existing(step_df.columns, ['is_saturated_global'])
sat_pct = float(step_df[sat_key].astype(bool).mean()) if sat_key else None

reward_key = pick_first_existing(interval_df.columns, ['global_interval_reward'])
reward_sum = float(pd.to_numeric(interval_df[reward_key], errors='coerce').fillna(0).sum()) if reward_key else None

angle_key = pick_first_existing(step_df.columns, ['pendulum_angle_raw', 'pendulum_angle_norm'])
u_key = pick_first_existing(step_df.columns, ['u_total_saturated', 'u_total'])

card = {
    'episode_id': EPISODE_ID,
    'n_steps': n_steps,
    'n_intervals': n_intervals,
    'terminated': terminated,
    'termination_reason': termination_reason,
    'reward_sum_intervals': reward_sum,
    'sat_pct_steps': sat_pct,
    'max_abs_angle': absmax(step_df[angle_key]) if angle_key else None,
    'max_abs_u_total': absmax(step_df[u_key]) if u_key else None,
}
card
```

## Caja 6 — Timeline del episodio (tabla mínima + 1 gráfico útil)

```python
cols = [
    'interval_id',
    'interval_start_step_idx',
    'interval_end_step_idx',
    'interval_n_steps',
    'terminated',
    'termination_reason',
    'global_interval_reward',
]
available = [c for c in cols if c in interval_df.columns]
timeline_df = interval_df[available].copy()

if 'interval_start_step_idx' in interval_df.columns and 'interval_end_step_idx' in interval_df.columns and sat_key:
    sat_pct_list = []
    for a, b in zip(interval_df['interval_start_step_idx'], interval_df['interval_end_step_idx']):
        try:
            a, b = int(a), int(b)
        except Exception:
            sat_pct_list.append(np.nan)
            continue
        a = max(0, min(a, len(step_df)-1))
        b = max(0, min(b, len(step_df)-1))
        if b < a:
            a, b = b, a
        sat_pct_list.append(float(step_df.loc[a:b, sat_key].astype(bool).mean()))
    timeline_df['sat_pct_interval'] = sat_pct_list

timeline_df.head(10)
```

```python
if angle_key is None:
    raise KeyError('No encuentro pendulum_angle_raw/norm en step_df. Revisa nombres.')
if u_key is None:
    raise KeyError('No encuentro u_total_saturated/u_total en step_df. Revisa nombres.')

x = np.arange(len(step_df))
fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(x, step_df[angle_key], label=angle_key)
ax.set_xlabel('step')
ax.set_ylabel('angle')
ax.grid(True)

ax2 = ax.twinx()
ax2.plot(x, step_df[u_key], alpha=0.6, label=u_key)
ax2.set_ylabel('u_total')

if 'interval_start_step_idx' in interval_df.columns:
    for s in pd.to_numeric(interval_df['interval_start_step_idx'], errors='coerce').dropna().astype(int).tolist():
        ax.axvline(s, linewidth=0.8, alpha=0.3)

if sat_key:
    sat_mask = step_df[sat_key].astype(bool).values
    ax.scatter(x[sat_mask], step_df.loc[sat_mask, angle_key], s=8, alpha=0.6)

lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper right')

plt.show()
```

## Caja 7 — Ledger del intervalo seleccionado (C1..C5 evidencia)

```python
if 'interval_id' not in interval_df.columns:
    raise KeyError('No existe interval_id en interval_df. Revisa nombres.')

matches = interval_df.index[interval_df['interval_id'] == INTERVAL_ID].tolist()
idx = matches[0] if matches else int(INTERVAL_ID)
row = interval_df.loc[idx].to_dict()

start_step = int(row.get('interval_start_step_idx', 0) or 0)
end_step = int(row.get('interval_end_step_idx', start_step) or start_step)
start_step = max(0, min(start_step, len(step_df)-1))
end_step = max(0, min(end_step, len(step_df)-1))
if end_step < start_step:
    start_step, end_step = end_step, start_step

start_step, end_step
```

```python
# C1: Contexto (estado + gains + flags) en start_step
state_keys_raw = ['cart_position_raw', 'cart_velocity_raw', 'pendulum_angle_raw', 'pendulum_velocity_raw']
state_keys_norm = ['cart_position_norm', 'cart_velocity_norm', 'pendulum_angle_norm', 'pendulum_velocity_norm']
state_keys = state_keys_raw if USE_RAW_STATE else state_keys_norm
state_keys = [k for k in state_keys if k in step_df.columns]
state_at_start = {k: step_df.loc[start_step, k] for k in state_keys}

gain_keys = [c for c in step_df.columns if isinstance(c, str) and c.startswith(('kp_', 'ki_', 'kd_'))]
gains_at_start = {k: step_df.loc[start_step, k] for k in gain_keys}

flags_at_start = {
    'is_saturated_global': step_df.loc[start_step, sat_key] if sat_key else None,
    'u_total_saturated': step_df.loc[start_step, 'u_total_saturated'] if 'u_total_saturated' in step_df.columns else None,
}

context = {
    'episode_id': EPISODE_ID,
    'interval_id': INTERVAL_ID,
    'interval_idx': idx,
    'start_step': start_step,
    'end_step': end_step,
    'state': state_at_start,
    'gains': gains_at_start,
    'flags': flags_at_start,
}
context
```

```python
# C2: Percepción RL (lo que existe en interval_data)
action_cols = [c for c in interval_df.columns if isinstance(c, str) and c.startswith('action_')]
td_cols = [c for c in interval_df.columns if isinstance(c, str) and c.startswith('td_error_')]
qv_cols = [c for c in interval_df.columns if isinstance(c, str) and c.startswith('q_value_')]

rl = {
    'epsilon': row.get('epsilon', None),
    'actions': {c: row.get(c) for c in action_cols},
    'td_errors': {c: row.get(c) for c in td_cols},
    'q_values': {c: row.get(c) for c in qv_cols},
    'state_idx': {c: row.get(c) for c in interval_df.columns if isinstance(c, str) and c.startswith('s_idx_')},
    'q_vec': {c: row.get(c) for c in interval_df.columns if isinstance(c, str) and c.startswith('q_vec_')},
    'is_explore': {c: row.get(c) for c in interval_df.columns if isinstance(c, str) and c.startswith('is_explore_')},
}
rl
```

```python
# C3: Intervención (delta gains entre (start_step-1) y start_step)
prev_step = max(0, start_step - 1)
gains_prev = {k: step_df.loc[prev_step, k] for k in gain_keys}
delta_gains = {}
for k in gain_keys:
    try:
        delta_gains[k] = float(gains_at_start[k]) - float(gains_prev[k])
    except Exception:
        delta_gains[k] = None

intervention = {
    'actions': rl['actions'],
    'gains_before': gains_prev,
    'gains_after': gains_at_start,
    'delta_gains': delta_gains,
}
intervention
```

```python
# C4: Ejecución (step-level) — muestras + ventanas por eventos
exec_cols = []
for k in ['t_sec'] + state_keys_raw + state_keys_norm:
    if k in step_df.columns:
        exec_cols.append(k)
for k in ['u_total', 'u_total_raw', 'u_total_saturated', 'delta_u_total', 'is_saturated_global']:
    if k in step_df.columns:
        exec_cols.append(k)
for k in ['u_contrib_pendulum_angle', 'u_contrib_cart_position']:
    if k in step_df.columns:
        exec_cols.append(k)
for pref in ['error_', 'derivative_error_', 'integral_error_', 'control_action_', 'delta_control_action_', 'is_saturated_', 'saturation_proportion_']:
    exec_cols += [c for c in step_df.columns if isinstance(c, str) and c.startswith(pref)]
exec_cols += gain_keys
exec_cols = list(dict.fromkeys(exec_cols))

interval_steps = step_df.loc[start_step:end_step, exec_cols].copy()

if len(interval_steps) <= N_UNIFORM:
    uniform = interval_steps
else:
    picks = np.linspace(start_step, end_step, N_UNIFORM).round().astype(int)
    uniform = step_df.loc[picks, exec_cols].copy()

angle_series = pd.to_numeric(step_df.loc[start_step:end_step, angle_key], errors='coerce')
peak_step = int(angle_series.abs().idxmax()) if angle_series.notna().any() else start_step

sat_start_step = None
if sat_key:
    sat_slice = step_df.loc[start_step:end_step, sat_key].astype(bool)
    idxs = sat_slice.index[sat_slice.values].tolist()
    sat_start_step = int(idxs[0]) if idxs else None

windows = {
    'start_window': step_df.loc[start_step:min(end_step, start_step+WINDOW_N), exec_cols].copy(),
    'end_window': step_df.loc[max(start_step, end_step-WINDOW_N):end_step, exec_cols].copy(),
    'around_peak_angle': step_df.loc[window_indices(peak_step, WINDOW_N, start_step, end_step), exec_cols].copy(),
}
if sat_start_step is not None:
    windows['around_saturation_start'] = step_df.loc[window_indices(sat_start_step, WINDOW_N, start_step, end_step), exec_cols].copy()

execution = {
    'uniform_samples': uniform.reset_index(drop=False).rename(columns={'index':'n_step'}),
    'windows': {k: v.reset_index(drop=False).rename(columns={'index':'n_step'}) for k, v in windows.items()},
}
execution['uniform_samples'].head(10)
```

```python
# Gráfico corto: ventana alrededor del pico de ángulo (angle + u_total)
w = execution['windows']['around_peak_angle'].copy()
x = w['n_step'].values if 'n_step' in w.columns else np.arange(len(w))

fig, ax = plt.subplots(figsize=(12, 3))
ax.plot(x, w[angle_key], label=angle_key)
ax.set_xlabel('step')
ax.set_ylabel('angle')
ax.grid(True)

ax2 = ax.twinx()
ax2.plot(x, w[u_key], alpha=0.6, label=u_key)
ax2.set_ylabel('u_total')

lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper right')
plt.show()
```

## Caja 8 — Evaluación (métricas → reward) + dominancia

```python
L_cols = [c for c in interval_df.columns if isinstance(c, str) and c.startswith('L_')]
evaluation = {
    'global_interval_reward': row.get('global_interval_reward', None),
    'controller_pendulum_angle_interval_reward': row.get('controller_pendulum_angle_interval_reward', None),
    'controller_cart_position_interval_reward': row.get('controller_cart_position_interval_reward', None),
    'terminated': row.get('terminated', None),
    'termination_reason': row.get('termination_reason', None),
    'L_top': topk_L(row, k=TOPK_L, prefix='L_'),
    'L_all': {c: row.get(c) for c in L_cols},
}
evaluation
```

## Caja 9 — Aprendizaje (TD / Q) (lo que existe en logs)

```python
learning = {
    'td_errors': rl['td_errors'],
    'q_values': rl['q_values'],
    'q_stats_end': {k: v for k, v in end_data.items() if isinstance(k, str) and re.match(r'^q_(mean|std|max|min)_', k)},
    'has_q_tables_snapshot': isinstance(end_data.get('q_tables'), dict) and len(end_data.get('q_tables', {})) > 0,
    'has_visit_counts_snapshot': isinstance(end_data.get('visit_counts'), dict) and len(end_data.get('visit_counts', {})) > 0,
}
learning
```

## Caja 10 — Ledger final (estructura canónica)

```python
ledger = {
    'context': context,
    'rl': rl,
    'intervention': intervention,
    'execution': execution,
    'evaluation': evaluation,
    'learning': learning,
}
ledger.keys()
```

## Caja 11 — Motivación mínima (solo con evidencia)

```python
anchors = {
    'start_step': start_step,
    'end_step': end_step,
    'peak_abs_angle_in_interval': float(angle_series.abs().max()) if angle_series.notna().any() else None,
    'sat_pct_interval': float(step_df.loc[start_step:end_step, sat_key].astype(bool).mean()) if sat_key else None,
    'global_interval_reward': evaluation['global_interval_reward'],
    'L_top': evaluation['L_top'],
}

because = []
if anchors['sat_pct_interval'] is not None and anchors['sat_pct_interval'] > 0.2:
    because.append('saturación relevante dentro del intervalo')
if anchors['peak_abs_angle_in_interval'] is not None:
    because.append(f"pico |angle|≈{anchors['peak_abs_angle_in_interval']:.4f}")

then = [f"{k}={v}" for k, v in rl['actions'].items()]
dominant = ', '.join([f"{x['name']}={x['value']}" for x in anchors['L_top']])

text = (
    f"Porque: {', '.join(because) if because else 'sin gatillantes obvios (solo evidencia base)'}\n"
    f"Entonces (acciones): {', '.join(then) if then else 'sin acciones registradas'}\n"
    f"Por eso (reward): global_interval_reward={anchors['global_interval_reward']}, dominancia={dominant}\n"
    f"Aprendizaje (TD): {learning['td_errors']}\n"
)
print(text)
```
